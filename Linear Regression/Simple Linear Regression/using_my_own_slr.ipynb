```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=200, 
                       n_features = 1, 
                       noise = 20, 
                       random_state=2)
```

```python
from sklearn.model_selection import train_test_split as trTest
X_train, X_test, y_train, y_test = trTest(X, y, 
                                            test_size=0.2,
                                            random_state = 2)
X_train
```

```python
from myOwnSLR_from_scratch import simpleLinearRegression 
reg = simpleLinearRegression()
reg.fit(X_train, y_train)
reg.coef_
```

```python
import matplotlib.pyplot as plt
plt.xlim(X[:, 0].min(), X[:, 0].max())
plt.ylim(y.min(), y.max())
plt.scatter(X, y, 
            color = 'blue',
            s=200,
            edgecolor='k',
            label = 'data__points')
plt.plot(X_train, 
         reg.predict(X_train), 
         color = 'black',
         linewidth = 5,
         label = "regression__line")
plt.legend()
plt.show()
```
