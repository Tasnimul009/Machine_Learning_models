```python
from sklearn.datasets import make_regression 
X, y = make_regression(n_samples = 200, 
                       n_features = 2, 
                       noise = 20, 
                       )
```

Making it a dataframe

```python
import pandas as pd
df = pd.DataFrame({'feature1' : X[:, 0],
                   'feature2' : X[:, 1],
                   'target' : y})
df
```

```python
from sklearn.model_selection import train_test_split as tts
X_train, X_test, y_train, y_test = tts(X, y, 
                                       test_size=0.2, 
                                       random_state=2)
```

```python
from myOwnMLR_scratch import multiple_linear_regression
reg = multiple_linear_regression()
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)
y_pred
```

```python
#visulise the plane 
import numpy as np 
from mpl_toolkits.mplot3d import Axes3D
X1_range = np.linspace(X[:, 0].min(), X[:, 0].max(), 50)
X2_range = np.linspace(X[:, 1].min(), X[:, 1].max(), 50)
X1_grid, X2_grid = np.meshgrid(X1_range, X2_range)
```

```python
#now y axis
y_pred = reg.predict(np.c_[X1_grid.ravel(), X2_grid.ravel()])
y_grid = y_pred.reshape(X1_grid.shape)
```

```python
import plotly.graph_objects as go 
import plotly.express as px
fig = px.scatter_3d(df, x = 'feature1', 
                    y = 'feature2', 
                    z = 'target')
fig.add_trace(go.Surface(x = X1_grid, 
                         y = X2_grid,
                         z = y_grid))
```
